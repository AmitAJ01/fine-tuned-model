# FineTuneOps — Baseline Evaluation on Cloud GPU
### Pretrained Baseline: `Qwen/Qwen2.5-Coder-1.5B-Instruct` on Cloud-OpsBench

This notebook evaluates the unmodified foundation model (`Qwen/Qwen2.5-Coder-1.5B-Instruct`) on the held-out Cloud-OpsBench test dataset (148 cases).

**Core Evaluation Principles:**
1. **Unmodified Weights**: Pretrained base model with zero adapters or fine-tuning.
2. **Zero Target Leakage**: Model inputs contain only system prompt and sanitized query/diagnostic traces.
3. **Deterministic Decoding**: Greedy decoding (`temperature=0.0`, `do_sample=False`).
4. **Benchmark Ground Truth**: Official ground truth includes Fault Taxonomy, Affected Component, and Root Cause. Derived severity and unprovided remediation are excluded from primary metrics.
